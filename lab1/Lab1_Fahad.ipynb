## Task 2 - Loading and Visualizing Images

```python
# prepare test images
import os, numpy as np
from skimage import data
from PIL import Image as PILImage

os.makedirs('images', exist_ok=True)
PILImage.fromarray(data.camera()).save('images/cameraman.tif')
coins = data.coins()
PILImage.fromarray(np.array(PILImage.fromarray(coins).resize((256,256)))).save('images/lena_gray_256.tif')
```

```python
import cv2
img = cv2.imread('images/cameraman.tif') #Load the image data into a NumPy array

import matplotlib.pyplot as plt
plt.imshow(img)
plt.show()
```

```python
from PIL import Image
img2 = Image.open('images/lena_gray_256.tif') #load image into a PIL Image object

import matplotlib.pyplot as plt
import matplotlib.cm as cm
plt.imshow(img2, cmap = cm.Greys_r)
plt.show()
```
