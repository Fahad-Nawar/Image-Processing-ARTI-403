### Task 1: Geometric Transformations

#### 1. Increase the size of the image

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('new_image.jpg')

# Scale the image to 2x its original size
scale_x = 2
scale_y = 2
resized_img = cv2.resize(img, None, fx=scale_x, fy=scale_y, interpolation=cv2.INTER_LINEAR)

print('Original size:', img.shape)
print('Resized size :', resized_img.shape)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
axes[0].set_title('Original Image')
axes[0].axis('off')

axes[1].imshow(cv2.cvtColor(resized_img, cv2.COLOR_BGR2RGB))
axes[1].set_title('Increased Size (2x)')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

#### 2. Rotate the image by 120 degrees

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('new_image.jpg')

h, w = img.shape[:2]
center = (w // 2, h // 2)

# Rotation matrix for 120 degrees
rotation_matrix = cv2.getRotationMatrix2D(center, 120, 1.0)
rotated_img = cv2.warpAffine(img, rotation_matrix, (w, h))

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
axes[0].set_title('Original Image')
axes[0].axis('off')

axes[1].imshow(cv2.cvtColor(rotated_img, cv2.COLOR_BGR2RGB))
axes[1].set_title('Rotated 120 Degrees')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

#### 3. Perform Shear operation

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('new_image.jpg')

h, w = img.shape[:2]

# Shear transformation matrix (horizontal shear)
shear_factor = 0.3
shear_matrix = np.float32([[1, shear_factor, 0],
                            [0, 1,            0]])

new_w = int(w + shear_factor * h)
sheared_img = cv2.warpAffine(img, shear_matrix, (new_w, h))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
axes[0].set_title('Original Image')
axes[0].axis('off')

axes[1].imshow(cv2.cvtColor(sheared_img, cv2.COLOR_BGR2RGB))
axes[1].set_title('Sheared Image (factor = 0.3)')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

### Task 2: Intensity Transformations

#### 1. Create a Negative Image

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('new_image2.jpg')
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Negative: subtract each pixel from 255
negative_img = 255 - gray_img

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(gray_img, cmap='gray')
axes[0].set_title('Original Image')
axes[0].axis('off')

axes[1].imshow(negative_img, cmap='gray')
axes[1].set_title('Negative Image')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

#### 2. Log Transformation

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('new_image2.jpg')
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Log transformation: s = c * log(1 + r)
# c is chosen so output scales to 0-255
c = 255 / np.log(1 + np.max(gray_img))
log_img = c * np.log(1 + gray_img.astype(np.float32))
log_img = np.uint8(log_img)

print('Constant c =', round(c, 4))

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(gray_img, cmap='gray')
axes[0].set_title('Original Image')
axes[0].axis('off')

axes[1].imshow(log_img, cmap='gray')
axes[1].set_title(f'Log Transformation (c = {round(c, 2)})')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

#### 3. Power Law (Gamma) Transformation

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('new_image2.jpg')
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Power law transformation: s = c * r^gamma
# gamma = 0.4 (< 1) brightens the image and increases contrast in dark regions
gamma = 0.4

normalized = gray_img.astype(np.float32) / 255.0
power_img = np.power(normalized, gamma)
power_img = np.uint8(power_img * 255)

print('Gamma =', gamma)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(gray_img, cmap='gray')
axes[0].set_title('Original Image')
axes[0].axis('off')

axes[1].imshow(power_img, cmap='gray')
axes[1].set_title(f'Power Law Transformation (gamma = {gamma})')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```
